## Local Inference on GPU
Model page: https://huggingface.co/allenai/olmOCR-7B-0725

⚠️ If the generated code snippets do not work, please open an issue on either the [model repo](https://huggingface.co/allenai/olmOCR-7B-0725)
			and/or on [huggingface.js](https://github.com/huggingface/huggingface.js/blob/main/packages/tasks/src/model-libraries-snippets.ts) 🙏

```python
from pydantic_settings import BaseSettings

from transformers import AutoProcessor, AutoModelForImageTextToText # AutoModelForVision2Seq
from transformers import pipeline
```

```python
class Settings(BaseSettings):
  model: str = "allenai/olmOCR-7B-0825"
```

```python
settings = Settings()
```

```python
messages = [
    {
        "role": "user",
        "content": [
            {"type": "image", "url": "https://huggingface.co/datasets/huggingface/documentation-images/resolve/main/p-blog/candy.JPG"},
            {"type": "text", "text": "What animal is on the candy?"}
        ]
    },
]
```

```python
# Use a pipeline as a high-level helper
pipe = pipeline("image-text-to-text", model=settings.model)
```

```python
pipe(text=messages)
```

```python
# Load model directly

processor = AutoProcessor.from_pretrained(settings.model)
model = AutoModelForImageTextToText.from_pretrained(settings.model) # AutoModelForVision2Seq
inputs = processor.apply_chat_template(
	messages,
	add_generation_prompt=True,
	tokenize=True,
	return_dict=True,
	return_tensors="pt",
).to(model.device)
```

```python
outputs = model.generate(**inputs, max_new_tokens=40)
```

```python
print(processor.decode(outputs[0][inputs["input_ids"].shape[-1]:]))
```

```python
# Use a pipeline as a high-level helper
from transformers import pipeline

pipe = pipeline("image-text-to-text", model="allenai/olmOCR-7B-0725")
messages = [
    {
        "role": "user",
        "content": [
            {"type": "image", "url": "https://huggingface.co/datasets/huggingface/documentation-images/resolve/main/p-blog/candy.JPG"},
            {"type": "text", "text": "What animal is on the candy?"}
        ]
    },
]
pipe(text=messages)
```

```python
# Load model directly
from transformers import AutoProcessor, AutoModelForVision2Seq

processor = AutoProcessor.from_pretrained("allenai/olmOCR-7B-0725")
model = AutoModelForVision2Seq.from_pretrained("allenai/olmOCR-7B-0725")
messages = [
    {
        "role": "user",
        "content": [
            {"type": "image", "url": "https://huggingface.co/datasets/huggingface/documentation-images/resolve/main/p-blog/candy.JPG"},
            {"type": "text", "text": "What animal is on the candy?"}
        ]
    },
]
inputs = processor.apply_chat_template(
	messages,
	add_generation_prompt=True,
	tokenize=True,
	return_dict=True,
	return_tensors="pt",
).to(model.device)

outputs = model.generate(**inputs, max_new_tokens=40)
print(processor.decode(outputs[0][inputs["input_ids"].shape[-1]:]))
```
